## Project Overview & Objectives
This notebook demonstrates exploratory data analysis (EDA), data wrangling, data preprocessing, and visualization for passenger bookings, route demand, gross revenue, and bus fleet occupancy across Tamil Nadu transit corridors (Erode, Coimbatore, Salem, Tirupur, Chennai, Bangalore, Madurai, Tiruchirappalli, Ooty).

### Step 1: Importing Required Libraries (Unit 2: Python Foundations)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 150
```

### Step 2: Ingesting the Public Transport Dataset (Unit 1 & Unit 2)

```python
df = pd.read_csv('data/public_transport_dataset.csv')
print(f"Dataset Dimensions: {df.shape[0]} rows x {df.shape[1]} columns")
df.head()
```

### Step 3: Data Cleaning & Preprocessing (Unit 4: Missing Values, Type Conversion & Feature Creation)

```python
# Checking for missing values
print("Missing value count per column:")
print(df.isnull().sum())

# Type conversions and feature engineering
df['booking_timestamp'] = pd.to_datetime(df['booking_timestamp'])
df['travel_date'] = pd.to_datetime(df['travel_date'])
df['departure_hour'] = pd.to_datetime(df['departure_time'], format='%H:%M').dt.hour
df['day_of_week'] = df['booking_timestamp'].dt.day_name()

df[['distance_km', 'fare_per_seat', 'total_amount', 'occupancy_rate_pct']].describe()
```

### Step 4: Visualizations (Unit 5: Exploratory Data Analysis & Charting)

```python
# 1. Route-Wise Passenger Demand
plt.figure(figsize=(10, 5))
route_demand = df[df['booking_status'] == 'CONFIRMED'].groupby('route_name')['seats_booked'].sum().sort_values(ascending=False)
sns.barplot(x=route_demand.values, y=route_demand.index, hue=route_demand.index, palette='Blues_r', legend=False)
plt.title('Route-Wise Passenger Travel Demand (VCET Corridor Analysis)', fontweight='bold')
plt.xlabel('Total Seats Booked')
plt.ylabel('Transit Route')
plt.show()
```

```python
# 2. Route-Wise Revenue Generation
plt.figure(figsize=(10, 5))
route_rev = df[df['booking_status'] == 'CONFIRMED'].groupby('route_name')['total_amount'].sum().sort_values(ascending=False)
sns.barplot(x=route_rev.values, y=route_rev.index, hue=route_rev.index, palette='Greens_r', legend=False)
plt.title('Gross Revenue Generated by Transit Corridor (₹)', fontweight='bold')
plt.xlabel('Total Revenue (INR)')
plt.ylabel('Transit Route')
plt.show()
```

```python
# 3. Seat Occupancy Rate Distribution by Bus Type (Box Plot)
plt.figure(figsize=(9, 5))
sns.boxplot(x='bus_type', y='occupancy_rate_pct', data=df, hue='bus_type', palette='Set2', legend=False)
plt.title('Bus Seat Occupancy Rate Distribution by Bus Category', fontweight='bold')
plt.xlabel('Bus Type')
plt.ylabel('Occupancy Rate (%)')
plt.show()
```

```python
# 4. Correlation Heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(df[['distance_km', 'fare_per_seat', 'seats_booked', 'total_amount', 'occupancy_rate_pct', 'passenger_age']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Transit Indicators', fontweight='bold')
plt.show()
```
